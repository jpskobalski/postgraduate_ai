### **Ejercicio 1.**
De 10 monedas hay 2 monedas falsas (cargadas), que tienen probabilidad 0.8 de mostrar cara al ser lanzadas. Si se toma una moneda al azar, se lanza 11 veces, y en todas ellas se obtiene cara, ¿qué es más probable: que la moneda elegida sea justa o que esté cargada? Calcular $P(\text{falsa} | 11 \text{ caras})$ y $P(\text{justa} | 11 \text{ caras})$.


**Eventos:**

$J$: la moneda es justa

$F$: a moneda es falsa/cargada

Al elegir una moneda al azar:

$$
P(J)=\frac{8}{10}=0.8
$$

$$
P(F)=\frac{2}{10}=0.2
$$

Ademas, tenemos $C^{11}$ el evento obtener 11 caras seguidas.

Si la moneda es justa, cada lanzamiento tiene probabilidad 0.5 de cara; si es falsa, 0.8.

$$
P(C^{11}\mid J)=(0.5)^{11}=\frac{1}{2048}=0.00048828125
$$

$$
P(C^{11}\mid F)=(0.8)^{11}=0.08589934592
$$

Aplicamos el teorema de Bayes:

$$
P(J\mid C^{11})=\frac{P(C^{11}\mid J)P(J)}{P(C^{11})}
$$

$$
P(J\mid C^{11})=\frac{(0.5)^{11}\cdot 0.8}{0.017570494184}
$$

$$
P(J\mid C^{11}) \approx \frac{0.000390625}{0.017570494184}
$$

$$
P(J\mid C^{11}) \approx 0.0222
$$

Por el contrario:

$$
P(F\mid C^{11})=\frac{P(C^{11}\mid F)P(F)}{P(C^{11})}
$$

$$
P(F\mid C^{11})=\frac{(0.8)^{11}\cdot 0.2}{0.017570494184}
$$

$$
P(F\mid C^{11}) \approx \frac{0.017179869184}{0.017570494184}
$$

$$
P(F\mid C^{11}) \approx 0.9778
$$

**Solucion:**
Dado que se observaron 11 caras consecutivas, es, evidentemente, más probable que la moneda sea falsa.

$$
P(\text{falsa}\mid 11\ \text{caras}) \approx 0.9778 \quad (97.78\%)
$$

$$
P(\text{justa}\mid 11\ \text{caras}) \approx 0.0222 \quad (2.22\%)
$$

### **Ejercicio 2.** 

Sean $X$ e $Y$ dos v.a. continuas con densidad conjunta:

$$
f_{X,Y}(x,y) =
\begin{cases}
K y & \text{si } 100x^2 \le y \le 4x \\
0 & \text{en otro caso}
\end{cases}
$$

**Soporte:**
De $100x^2 \le y \le 4x$ se obtiene
$$
0 \le x \le \tfrac{1}{25}, \qquad 0 \le y \le \tfrac{4}{25},
$$
y, fijado $y \in (0, \tfrac{4}{25}]$,
$$
\tfrac{y}{4} \le x \le \tfrac{\sqrt{y}}{10}
$$




#### **a. Determine el valor de $K$**

Se plantea la integral doble de la ecuacion presentada, dado que asi nos permite despejar el valor de la constante K. 

$$
\int_{0}^{1/25} \int_{100x^2}^{4x} K y \, dy\, dx = 1
$$

$$
K \int_{0}^{1/25} \int_{100x^2}^{4x} y \, dy\, dx = 1
$$

Resolvemos la integral interna respecto de $y$:

$$
\int_{100x^2}^{4x} y \, dy
= \left.\frac{y^2}{2}\right|_{100x^2}^{4x}
= \frac{(4x)^2 - (100x^2)^2}{2}
$$

$$
= \frac{16x^2 - 10000x^4}{2}
= 8x^2 - 5000x^4
$$

Sustiuimos el resultado en la integral externa:

$$
K \int_{0}^{1/25} \big(8x^2 - 5000x^4\big) dx = 1
$$

$$
\int 8x^2 \, dx = \tfrac{8}{3}x^3, 
\qquad
\int 5000x^4 \, dx = 1000x^5
$$

Entonces:

$$
K \left[ \tfrac{8}{3}x^3 - 1000x^5 \right]_{0}^{1/25} = 1
$$

Evaluamos en $x = \tfrac{1}{25}$:

$$
\tfrac{8}{3}\left(\tfrac{1}{25}\right)^3 - 1000\left(\tfrac{1}{25}\right)^5
= \tfrac{8}{46875} - \tfrac{1000}{9765625}
= \tfrac{16}{234375}
$$


$$
\boxed{K = \frac{234375}{16}}
$$


#### **b. Densidad marginal de $Y$**

Se integra la densidad conjunta con respecto a la variable x, ya que esto nos permite eliminarla y obtener la funcion dependiente solo de y. Para ello primero se determina el intervalo valido de integracion en funcion de y, y luego se calcula la integral, llegando a la expresion de f_Y(y).


Para cada $y \in (0,\tfrac{4}{25}]$, el intervalo valido de $x$ es $\big[\tfrac{y}{4}, \tfrac{\sqrt{y}}{10}\big]$

$$
f_Y(y)
=\int_{y/4}^{\sqrt{y}/10} K\,y\,dx
=K\,y\left(\tfrac{\sqrt{y}}{10}-\tfrac{y}{4}\right),
\qquad 0<y\le \tfrac{4}{25}.
$$

$K=\tfrac{234375}{16}$ 

$$
f_Y(y)
=K\left(\frac{y^{3/2}}{10}-\frac{y^{2}}{4}\right)
=\frac{234375}{16}\cdot\frac{y^{3/2}}{10}\;-\;\frac{234375}{16}\cdot\frac{y^{2}}{4}
$$

$$
\frac{234375}{16}\cdot\frac{1}{10}=\frac{234375}{160}=\frac{46875}{32},
\qquad
\frac{234375}{16}\cdot\frac{1}{4}=\frac{234375}{64}
$$

$$
\boxed{
f_Y(y)=\frac{46875}{32}\,y^{3/2}-\frac{234375}{64}\,y^{2},
\qquad 0<y\le \tfrac{4}{25},
\quad\text{y}\quad f_Y(y)=0 \;\text{en otro caso}}
$$

#### **c) Densidad condicional $f_{X|Y}(x|y)$**

Se aplica la definicion de densidad condicional, dividiendo la densidad conjunta entre la marginal de Y. Al sustituir, podemos observar que la expresion se simplifica y resulta una densidad uniforme en el intervalo valido de x para cada y. Esto muestra que X, dado Y=y, se distribuye uniformemente en ese rango.

$$
f_{X|Y}(x|y)=\frac{f_{X,Y}(x,y)}{f_Y(y)}.
$$

Para cada $y \in (0,\tfrac{4}{25}]$, el intervalo válido de $x$ es 
$$
\Big[\tfrac{y}{4}, \tfrac{\sqrt{y}}{10}\Big].
$$

Sabemos que 
$$
f_{X,Y}(x,y)=K\,y, \quad \text{si } \tfrac{y}{4}\le x \le \tfrac{\sqrt{y}}{10},
$$
y que
$$
f_Y(y)=K\,y\Bigl(\tfrac{\sqrt{y}}{10}-\tfrac{y}{4}\Bigr).
$$

Entonces,
$$
f_{X|Y}(x|y)=\frac{Ky}{Ky\left(\tfrac{\sqrt{y}}{10}-\tfrac{y}{4}\right)}
=\frac{1}{\tfrac{\sqrt{y}}{10}-\tfrac{y}{4}},
\qquad \tfrac{y}{4}\le x \le \tfrac{\sqrt{y}}{10}, \quad 0<y\le \tfrac{4}{25}.
$$

Notamos que $X|Y=y$ es una uniforme sobre el intervalo válido, pues la densidad es constante y se anula fuera de él.

$$
\boxed{
f_{X|Y}(x|y)=
\begin{cases}
\dfrac{1}{\tfrac{\sqrt{y}}{10}-\tfrac{y}{4}}, & \tfrac{y}{4}\le x \le \tfrac{\sqrt{y}}{10}, \; 0<y\le \tfrac{4}{25}, \\[8pt]
0, & \text{en otro caso}.
\end{cases}}
$$

### **Ejercicio 3.**

Don Francisco es un pequeno comerciante de barrio que posee un supermercado, con el que sostiene su familia. 
Uno de sus hijos, Matıas, quien recien inicia a cursar la Especializacion en Inteligencia Artificial del LSE de la UBA, le
propone hacer un analisis de las ventas durante el ano anterior, con el fin de hacer pronosticos para el ano siguiente, lo que 
a don Francisco le parece buena idea.
Don Francisco le entrega a Matıas el cuaderno donde tiene registrado el valor total de sus ventas en cada dıa del ano. Con 
esta informacion, Matıas construye una tabla en la cual la primera columna corresponde a la fecha y la segunda corresponde
al monto de las ventas, en dolares para evitarse dolores de cabeza con la inflacion. Matıas no se siente muy seguro de la 
tarea a realizar, ası que les pide ayuda a ustedes para abordar el problema. 
A partir del archivo de datos correspondiente a su grupo, determine una funcion empırica de distribucion y una aproximacion
a la funcion de densidad de las ventas del supermercado de Don Francisco para cada ano de registro (2021, 2022 y 2023):


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
"""
Comienzo la resolucion del enunciado. Solo quiero dejar una nota, no tengo la letra ñ en el teclado y puede ser que se me escape en algun lado.
"""

#defino el tamaño de los graficos
plt.rcParams["figure.figsize"] = (10, 5)

# Creo un dataframe en pandas para poder trabajar con los datos, se usa openpyxl
df = pd.read_excel("Datos_primer_TP_23Co2025_a2325.xlsx")

# Transformo las fechas importadas en datetime y creo una columna nueva, año, para poder trabajar mas adelante
df["Fecha"] = pd.to_datetime(df["Fecha"])
df["Año"] = df["Fecha"].dt.year

# filtro por los 3 años (por si existen otros datos en el excel)
df = df[df["Año"].isin([2021, 2022, 2023])].copy()

def distribucion_acumulada_empirica(valores):
    valores_ordenados = np.sort(valores)
    n = valores_ordenados.size
    probabilidad = np.arange(1, n + 1) / n
    return valores_ordenados, probabilidad

# recorro todos los años que tenemos en la data e imprimo el mismo set de graficos para cada uno de ellos
for ano in sorted(df["Año"].unique()):
    datos = df.loc[df["Año"] == ano, "Ventas"].dropna().to_numpy()

    # Funcion de Distribucion Acumulada Empirica
    x, y = distribucion_acumulada_empirica(datos)
    plt.figure()
    plt.step(x, y, where="post")
    plt.title(f"Distribucion Acumulada Empirica de Ventas - {ano}")
    plt.xlabel("Ventas ($)")
    plt.ylabel("F(x)")
    plt.tight_layout()
    plt.show()

    # Densidad e Histograma 
    plt.figure()
    # Histograma
    sns.histplot(datos, stat="density", bins=30, alpha=0.35)
    sns.kdeplot(datos, bw_adjust=1.0, fill=True, alpha=0.25)
    plt.title(f"Funcion de Densidad - {ano}")
    plt.xlabel("Ventas ($)")
    plt.ylabel("Densidad")
    plt.tight_layout()
    plt.show()
```
