### Ejercicio 1

Una variable aleatoria discreta $X$ puede tomar los valores $0,1,2,3$ con probabilidades:

$P(X=0)=\theta, \quad$ $P(X=1)=2\theta, \quad$ $P(X=2)=\frac{1-3\theta}{3}, \quad$ $P(X=3)=\frac{2(1-3\theta)}{3}.$

Restricción sobre $\theta$ Para que sea una distribución válida, cada probabilidad debe ser no negativa:

- $P(X=0)=\theta \geq 0 \;\;\Rightarrow\;\; \theta \geq 0$.
- $P(X=1)=2\theta \geq 0$ (ya vale si $\theta \geq 0$).
- $P(X=2)=\frac{1-3\theta}{3} \geq 0 \;\;\Rightarrow\;\; \theta \leq \tfrac{1}{3}$.
- $P(X=3)=\frac{2(1-3\theta)}{3} \geq 0 \;\;\Rightarrow\;\; \theta \leq \tfrac{1}{3}$.

Entonces podemos decir que $0 \leq \theta \leq \tfrac{1}{3}$

La suma de probabilidades también se cumple: $\theta + 2\theta + \frac{1-3\theta}{3} + \frac{2(1-3\theta)}{3} = 1$

Verosimilitud y estimador MV

Los datos observados son: $(1,1,3,0,3,1,2,2,3,2)$

Entonces las frecuencias: $n_0=1, \quad n_1=3, \quad n_2=3, \quad n_3=3, \quad n=10.$

La verosimilitud (sin constantes) es:

$$
L(\theta)\propto 
\theta^{n_0}(2\theta)^{n_1}\left(\tfrac{1-3\theta}{3}\right)^{n_2}
\left(\tfrac{2(1-3\theta)}{3}\right)^{n_3}.
$$

Simplificando:

$$
L(\theta) \propto \theta^{n_0+n_1}\,(1-3\theta)^{n_2+n_3}.
$$

Sea $a=n_0+n_1=4$, $b=n_2+n_3=6$.  
Si le aplicamos logaritmo:

$$
\ell(\theta) = a\ln\theta + b\ln(1-3\theta).
$$

Derivando e igualando a cero:

$$
\ell'(\theta) = \frac{a}{\theta} - \frac{3b}{1-3\theta} = 0
\;\;\Rightarrow\;\;
a(1-3\theta)=3b\theta
\;\;\Rightarrow\;\;
\hat{\theta} = \frac{a}{3(a+b)}.
$$

Con los datos:

$$
\boxed{
\hat{\theta} = \frac{4}{3\cdot 10} = \frac{2}{15} \approx 0.1333.}
$$

### Ejercicio 2

Se observa la relacion entre tiempo $X$ (meses) y producción $Y$ (miles de toneladas) en los puntos:

$$
(0,4),\ (7,17),\ (13,84),\ (17,251),\ (20,531).
$$

Se propone el modelo de regresion cuadratica
$$
Y = a + bX + cX^2.
$$

Para minimos cuadrados,
$$
\begin{pmatrix}
n & \sum X & \sum X^2\\
\sum X & \sum X^2 & \sum X^3\\
\sum X^2 & \sum X^3 & \sum X^4
\end{pmatrix}
\begin{pmatrix} a\\ b\\ c \end{pmatrix}
=
\begin{pmatrix}
\sum Y\\ \sum XY\\ \sum X^2Y
\end{pmatrix}.
$$

Con los datos:
$$
\begin{aligned}
&n=5,\quad \sum X = 57,\quad \sum X^2 = 907,\quad \sum X^3 = 15453,\quad \sum X^4 = 274{,}483,\\
&\sum Y = 887,\quad \sum XY = 16{,}098,\quad \sum X^2Y = 299{,}968.
\end{aligned}
$$

Entonces el sistema queda:
$$
\begin{pmatrix}
5 & 57 & 907\\
57 & 907 & 15453\\
907 & 15453 & 274483
\end{pmatrix}
\begin{pmatrix} a\\ b\\ c \end{pmatrix}
=
\begin{pmatrix}
887\\ 16098\\ 299968
\end{pmatrix}.
$$

Resolviendo el sistema (o equivalentemente $\hat\beta=(X^\top X)^{-1}X^\top y$ con $X=[\mathbf{1}, X, X^2]$):

$$
\boxed{
\hat a \approx 21.3723,\quad
\hat b \approx -24.7624,\quad
\hat c \approx 2.4163.}
$$

El modelo ajustado es:
$$
\hat Y \;=\; 21.3723\;-\;24.7624\,X\;+\;2.4163\,X^2.
$$

### Ejercicio 3

Datos: en 10 clientes, 2 estan en mora. Sea $p$ el porcentaje (probabilidad) de morosidad.

Modelo:
$$
Y \mid p \sim \mathrm{Binomial}(n=10,\;p), \qquad \text{dato observado: } y=2.
$$

A priori:
$$
p \sim \mathrm{Beta}(\alpha,\beta), \quad \alpha=1,\; \beta=4.
$$

Posterior (el cuadr de conjugacion no dice --> Beta–Binomial):
$$
p \mid y \sim \mathrm{Beta}(\alpha+y,\;\beta+n-y)
= \mathrm{Beta}(1+2,\;4+10-2)
= \boxed{\mathrm{Beta}(3,\,12)}.
$$

Media posterior:
$$
\mathbb{E}[p \mid y] = \frac{\alpha+y}{\alpha+\beta+n}
= \frac{3}{3+12} = \boxed{0.2}.
$$

Varianza posterior:
$$
\mathrm{Var}(p \mid y) =
\frac{(\alpha+y)(\beta+n-y)}{\big(\alpha+\beta+n\big)^2 \big(\alpha+\beta+n+1\big)}
= \frac{3\cdot 12}{15^2 \cdot 16}
= \boxed{0.01}.
$$

Entonces, podemos decir que:
- La distribucion a posteriori es $\mathrm{Beta}(3,12)$.
- El valor esperado posterior del porcentaje de morosidad es $0.20$ (20%).
- La varianza posterior es $0.01$.
